```python
df['Features'] = df['Genre'] + ' ' + df['Description']

print("MOVIE FEATURES CREATED SUCCESSFULLY!")
display(df[['Movie_Title', 'Features']])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['Features'])

print("TF-IDF conversion completed successfully!")
print("Number of movies:", tfidf_matrix.shape[0])
print("Number of features:", tfidf_matrix.shape[1])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(tfidf_matrix)

print("Movie similarity calculated successfully!")
print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
movie_title = 'Harry Potter'

movie_index = df[df['Movie_Title'] == movie_title].index[0]

scores = list(enumerate(similarity_matrix[movie_index]))
scores = sorted(scores, key=lambda x: x[1], reverse=True)

for index, score in scores[1:6]:
    print(df.iloc[index]['Movie_Title'], "-", df.iloc[index]['Genre'])
```

```python
import gradio as gr

def get_movies(movie_title):
    movie_index = df[df['Movie_Title'] == movie_title].index[0]
scores = list(enumerate(similarity_matrix[movie_index]))
scores = sorted(scores, key=lambda x: x[1], reverse=True)
result = [df.iloc[i]['Movie_Title'] for i, score in scores[1:6]]
print("\n".join(result))

app = gr.Interface(
                        fn=lambda movie_title: "\n".join(
                                [df.iloc[i]['Movie_Title'] for i, score in
                                         sorted(enumerate(similarity_matrix[
                                                      df[df['Movie_Title'] == movie_title].index[0]
                                                               ]), key=lambda x: x[1], reverse=True)[1:6]]
                                                                   ),
                                                                       inputs=gr.Dropdown(
                                                                               choices=df['Movie_Title'].tolist(),
                                                                                       label="Select a Movie"
                                                                                           ),
                                                                                               outputs=gr.Textbox(label="Recommended Movies"),
                                                                                                   title="Movie Recommendation System",
                                                                                                       description="Select a movie to discover five similar movies!"
                                                                                                       )

app.launch(share=True)
```
